```python
import pandas as pd

stats_df = pd.read_csv("astubbin_project_stats.csv", sep=";")
df = pd.read_csv("astubbin_project_summary.csv", sep=";", names=["project", "commit", "author", "time", "message"],)

df['Month'] = pd.to_datetime(df['time'], unit='s').dt.strftime('%Y-%m')

all_gap_commits = []

for _, row in stats_df.iterrows():
    project_name = row['Project']
    gap_start = row['LongestGapStart'] 
    gap_end = row['LongestGapEnd']   
    
    proj_df = df[df['project'] == project_name].sort_values(by='time').reset_index(drop=True)
    if proj_df.empty:
        continue
        
    pre_df = proj_df[proj_df['Month'] <= gap_start]
    last_10_pre = pre_df.tail(10).copy()
    last_10_pre.insert(0, 'pre_post', 'pre')
    
    post_df = proj_df[proj_df['Month'] >= gap_end]
    first_10_post = post_df.head(10).copy()
    first_10_post.insert(0, 'pre_post', 'post')
    
    last_10_pre = last_10_pre.drop(columns=['Month'])
    first_10_post = first_10_post.drop(columns=['Month'])
    
    all_gap_commits.append(pd.concat([last_10_pre, first_10_post]))

if all_gap_commits:
    final_df = pd.concat(all_gap_commits, ignore_index=True)
    final_df.to_csv("astubbin_project_gap_commits.csv", sep=";", index=False, header=False)
```
